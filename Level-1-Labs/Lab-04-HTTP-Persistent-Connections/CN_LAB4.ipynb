```python
import matplotlib.pyplot as plt

# Data from your recorded observations
connections = [0, 2, 4, 6, 8, 10]
# Calculated Load Times (End Time - Start Time)
load_times = [0.421448, 0.581529, 0.667445, 0.550696, 0.625826, 0.952906]

# Create the plot
plt.plot(connections, load_times, marker='o', linestyle='-', color='b', linewidth=2)

# Labeling according to lab manual specifications
plt.xlabel('Number of Persistent Connections') # [cite: 1038]
plt.ylabel('Total Page Load Time (seconds)')   # [cite: 1038]
plt.title('Total Page Load Time vs. Number of Persistent Connections')

# Add a grid for better readability
plt.grid(True, linestyle='--', alpha=0.7)

# Display the values on the points
for i, txt in enumerate(load_times):
    plt.annotate(f"{txt:.3f}s", (connections[i], load_times[i]), textcoords="offset points", xytext=(0,10), ha='center')

# Save the plot for submission
plt.savefig('http_experiment_plot.png')
print("Graph generated: http_experiment_plot.png")
```
